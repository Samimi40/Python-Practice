تعریف:

برنامه‌نویسی شی‌گرا (OOP) یک پارادایم برنامه‌نویسی مبتنی بر مفهوم "اشیاء" است که شامل داده‌ها (به صورت فیلدها یا ویژگی‌ها) و کد (به صورت رویه‌ها یا متدها) می‌باشند. در این پارادایم، برنامه‌ها با استفاده از برهم‌کنش اشیاء طراحی می‌شوند، به جای توابع و منطق خطی

Object-oriented programming (OOP) is a programming paradigm based on the concept of objects, which can contain data (fields/attributes) and code (procedures/methods). A feature of objects is that an object's own procedures can access and often modify the data fields of itself.

```python
class Student:
    def __init__(self, name, age, grade):
        # سازنده: وقتی شیء ساخته می‌شه اجرا میشه
        print(f" ثبت دانش‌آموز: {name}")
        self.name = name
        self.age = age
        self.grade = grade
        self.attendance = 0

    def present(self):
        self.attendance += 1
        print(f" {self.name} حاضر است (حضور: {self.attendance})")

# ساخت اشیاء (سازنده اجرا میشه)
student1 = Student("احمد", 12, "هفتم")  # ثبت دانش‌آموز: احمد
student2 = Student("مریم", 11, "ششم")   # ثبت دانش‌آموز: مریم

student1.present()  # احمد حاضر است (حضور: 1)
student1.present()  # احمد حاضر است (حضور: 2)
```

```python
class BreadMaker:
    def __init__(self, brand, color):
        self.brand = brand      # ویژگی: برند
        self.color = color      # ویژگی: رنگ
        self.is_on = False      # ویژگی: وضعیت روشن/خاموش

    def turn_on(self):          # متد: روشن کردن
        self.is_on = True
        print("دستگاه روشن شد!")

    def bake(self):            # متد: پخت نان
        if self.is_on:
            print(" نان تازه پخته شد!")
        else:
            print(" ابتدا دستگاه را روشن کن!")

# ایجاد شیء (نمونه از کلاس)
my_maker = BreadMaker("Xiaomi", "نقره‌ای")  # شیء اول
your_maker = BreadMaker("Philips", "سفید")  # شیء دوم

# استفاده از اشیاء
my_maker.turn_on()   # دستگاه روشن شد!
my_maker.bake()      #  نان تازه پخته شد!

your_maker.bake()    #  ابتدا دستگاه را روشن کن!
```

```python
class Student:
    def __init__(self, name, student_id, courses=None):
        self.name = name
        self.student_id = student_id
        self.courses = courses if courses else []
        self.gpa = 0.0

    def add_course(self, course):
        self.courses.append(course)
        print(f"درس {course} به {self.name} اضافه شد.")

    def set_gpa(self, gpa):
        if 0 <= gpa <= 20:
            self.gpa = gpa
            print(f"معدل {self.name} به {gpa} تغییر کرد.")
        else:
            print("معدل نامعتبر!")

    def show_info(self):
        print(f"\nدانشجو: {self.name}")
        print(f"شماره: {self.student_id}")
        print(f"دروس: {', '.join(self.courses) if self.courses else 'هیچ'}")
        print(f"معدل: {self.gpa}")

amir = Student("امیر", "98123456")
amir.add_course("ریاضی")
amir.add_course("برنامه‌نویسی")
amir.set_gpa(17.5)
amir.show_info()
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print(f"واریز {amount} - موجودی: {self.__balance}")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"برداشت {amount} - موجودی: {self.__balance}")
        else:
            print("موجودی ناکافی!")

    def show_balance(self):
        print(f"موجودی {self.owner}: {self.__balance}")

account = BankAccount("امیر", 1000000)
account.deposit(500000)
account.withdraw(300000)
account.show_balance()

print("\n" + "^"*44)
print(" تمام بخش‌ها با موفقیت اجرا شدند!")
```

```python
#موجودی فقط در صورت واریز و برداشت تغییر کند
class BankAccount:
    def __init__(self, initial_balance):
        self.__balance = initial_balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"{amount} واریز شد. موجودی: {self.__balance}")
        else:
            print("مبلغ واریز باید مثبت باشد!")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"{amount} برداشت شد. موجودی: {self.__balance}")
        else:
            print("مبلغ نامعتبر یا موجودی ناکافی!")

acc1 = BankAccount(1000)
acc1.deposit(500)    # 500 واریز شد. موجودی: 1500
acc1.withdraw(200)   # 200 برداشت شد. موجودی: 1300
acc1.withdraw(2000)  # خطا
```

```python
#نمره نباید خارج از بازه 0 تا 20 باشد
class Student:
    def __init__(self, name):
        self.name = name
        self.__scores = []

    def add_score(self, score):
        if 0 <= score <= 20:
            self.__scores.append(score)
            print(f"نمره {score} اضافه شد")
        else:
            print("نمره باید بین 0 تا 20 باشد!")

    def get_average(self):
        if not self.__scores:
            return 0
        return sum(self.__scores) / len(self.__scores)

s = Student("Amir")
s.add_score(18)   # نمره 18 اضافه شد
s.add_score(25)   # خطا: نمره باید بین 0 تا 20 باشد!
s.add_score(14)
print(s.get_average())  # 16.0
```

```python
#یک سیستم کوچک(هتل) که تعداد اتاق‌های موجود را مخفی می‌کند و فقط از طریق رزرو و لغو رزرو تغییر می‌کند

class Hotel:
    def __init__(self, total_rooms):
        self.__available_rooms = total_rooms
        self.__reserved_rooms = 0

    def get_available_rooms(self):
        return self.__available_rooms

    def reserve_room(self):
        if self.__available_rooms > 0:
            self.__available_rooms -= 1
            self.__reserved_rooms += 1
            print(f"رزرو انجام شد. اتاق‌های باقی‌مانده: {self.__available_rooms}")
        else:
            print("متاسفانه اتاق خالی وجود ندارد!")

    def cancel_reservation(self):
        if self.__reserved_rooms > 0:
            self.__available_rooms += 1
            self.__reserved_rooms -= 1
            print(f"رزرو لغو شد. اتاق‌های موجود: {self.__available_rooms}")
        else:
            print("هیچ رزروی برای لغو وجود ندارد!")

hotel = Hotel(5)
hotel.reserve_room()   # رزرو انجام شد. اتاق‌های باقی‌مانده: 4
hotel.reserve_room()   # رزرو انجام شد. اتاق‌های باقی‌مانده: 3
hotel.cancel_reservation()  # رزرو لغو شد. اتاق‌های موجود: 4
hotel.reserve_room()   # رزرو انجام شد. اتاق‌های باقی‌مانده: 3
```

# Inheritance

 ارث‌ بری به شما این امکان را می‌دهد که یک کلاس را تازه‌سازی کنید (که به آن کلاس فرزند یا کلاس مشتق‌شده می‌گوییم) که دقیقاً بر اساس یک کلاس از پیش نوشته‌شده (که به آن کلاس والد یا کلاس پایه می‌گوییم) ساخته می‌شود.

وقتی یک کلاس فرزند از کلاس والد خود ارث‌می‌برد، یعنی:

همهٔ ویژگی‌ها (همان متغیرهای درون کلاس) و

همهٔ کارکردها (همان توابعی که درون کلاس تعریف شده‌اند)

کلاس والد را به ارث می‌برد؛ یعنی بدون این که دوباره آن‌ها را بنویسد، در اختیار دارد. بعد از آن، کلاس فرزند می‌تواند این موارد را به سه شکل تغییر دهد:



   * آن‌ها را بازنویسی یا تغییر دهد (کارکرد والد را مطابق نیاز خودش عوض کند)

داشته باشد).


1.   از آن‌ها استفاده کند (دست بزند به آن‌ها)
2.    آن‌ها را بازنویسی یا تغییر دهد (کارکرد والد را مطابق نیاز خودش عوض کند).

3.    ویژگی‌ها و کارکردهای تازه‌ای به آن اضافه کند (چیزی که در والد نبوده را خودش داشته باشد)

 ازاین برای صدا زدن سازنده‌ی والد استفاده می‌کنیم تا مطمئن شویم ویژگی‌های والد به درستی ساخته می‌شوند.





```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def info(self):
        return f"Name: {self.name}, Age: {self.age}"

class Dog(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)  # صدا زدن سازنده والد
        self.breed = breed  # ویژگی جدید

    def info(self):  # بازنویسی
        return f"{super().info()}, Breed: {self.breed}"

d = Dog("Buddy", 3, "Golden Retriever")
print(d.info())  # Name: Buddy, Age: 3, Breed: Golden Retriever
```

```python
#یک کلاس والد، یک کلاس فرزند، و یک نوه (کلاس نتیجه).

class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def start(self):
        return f"{self.brand} vehicle is starting"

class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors

    def honk(self):
        return f"{self.brand} car says: Beep Beep!"

class ElectricCar(Car):  # ارث‌بری از Car که خودش از Vehicle ارث برده
    def __init__(self, brand, doors, battery_capacity):
        super().__init__(brand, doors)
        self.battery_capacity = battery_capacity

    def start(self):  # بازنویسی متد start
        return f"{self.brand} electric car starts silently!"

tesla = ElectricCar("Tesla", 4, 75)
print(tesla.start())   # Tesla electric car starts silently! (بازنویسی شده)
print(tesla.honk())    # Tesla car says: Beep Beep! (از Car به ارث برده)
```

```python
#کلاس Employee و دو فرزند Manager و Developer با محاسبه‌ی حقوق متفاوت

class Employee:
    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary

    def get_salary(self):
        return self.base_salary

    def work(self):
        return f"{self.name} is working"

class Manager(Employee):
    def __init__(self, name, base_salary, bonus):
        super().__init__(name, base_salary)
        self.bonus = bonus

    def get_salary(self):  # بازنویسی
        return self.base_salary + self.bonus

    def manage_team(self):  # متد جدید
        return f"{self.name} is managing team"

class Developer(Employee):
    def __init__(self, name, base_salary, language):
        super().__init__(name, base_salary)
        self.language = language

    def work(self):
        return f"{self.name} is writing code in {self.language}"

m = Manager("Sara", 5000, 1000)
d = Developer("Amir", 4000, "Python")
print(m.get_salary())  # 6000
print(m.work())        # Sara is working (از والد ارث برده)
print(d.work())        # Amir is writing code in Python (بازنویسی شده)
```

چندریختی در پایتون یک پارادایم برنامه‌نویسی است که در آن:


*   یک عملکرد واحد (تابع، متد یا عملگر) قادر است بر روی انواع مختلف داده عمل کند
*   تصمیم‌گیری درباره نحوه اجرا بر اساس رفتار شیء در زمان اجرا انجام می‌شود
*   رابط برنامه‌نویسی از جزئیات پیاده‌سازی جدا شده است


Polymorphism in Python is the provision of a single interface to entities of different types, or the use of a single symbol to represent multiple different types. Python implements polymorphism through duck typing, where the type of an object is determined by its behavior (methods and properties) rather than its explicit inheritance hierarchy

```python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.speak())
```

# Abstraction


انتزاع در پایتون، یک اصل بنیادی برنامه‌نویسی شی‌گرا است که به برنامه‌نویس اجازه می‌دهد پیچیدگی‌های سیستم را با تعریف رابط‌های ساده برای اشیاء پیچیده مدیریت کند. در این روش، جزئیات غیرضروری از دید کاربر نهایی پنهان می‌شوند و فقط عملکردهای اصلی در معرض دید قرار می‌گیرند. پایتون این مفهوم را از طریق کلاس‌های پایه انتزاعی و اجبار به پیاده‌سازی متدهای خاص در کلاس‌های فرزند پیاده‌سازی می‌کند

Abstraction in Python is a fundamental principle of object-oriented programming that allows programmers to manage system complexity by defining simple interfaces for complex objects. In this approach, unnecessary details are hidden from the end-user, and only core functionalities are exposed. Python implements this concept through abstract base classes and enforcing the implementation of specific methods in child classes.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass  # فقط قرارداد است، هیچ کدی نداره

class Dog(Animal):
    def speak(self):  # اجباری: باید بنویسمش
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

dog = Dog()
cat = Cat()
print(dog.speak())  # Woof!
print(cat.speak())  # Meow!

#  این خطا می‌دهد (نمی‌شه از کلاس انتزاعی نمونه ساخت):
# animal = Animal()  # TypeError!
```

```python
from abc import ABC, abstractmethod

class WritingTool(ABC):
    @abstractmethod
    def write(self):
        pass

class Pen(WritingTool):
    def write(self):
        return "Writing with pen"

class Pencil(WritingTool):
    def write(self):
        return "Writing with pencil"

pen = Pen()
pencil = Pencil()
print(pen.write())
print(pencil.write())
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def speak(self):
        pass

    def show_name(self):  # متد معمولی (همه ارث می‌برن)
        return f"My name is {self.name}"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"


dog = Dog("Oddy")
cat = Cat("Kitty")
print(dog.show_name())
print(dog.speak())
print(cat.speak())
```

```python
from abc import ABC, abstractmethod

class Instrument(ABC):
    @abstractmethod
    def play(self):
        pass

    def tune(self):  # متد معمولی
        return "Instrument is tuned"

class Guitar(Instrument):
    def play(self):
        return "Strumming guitar"

class Piano(Instrument):
    def play(self):
        return "Playing piano"


guitar = Guitar()
piano = Piano()
print(guitar.tune())  # Instrument is tuned
print(guitar.play())  # Strumming guitar
print(piano.play())   # Playing piano
```

```python
from abc import ABC, abstractmethod

class Reservation(ABC):
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.is_booked = False

    @abstractmethod
    def book(self):
        pass

    @abstractmethod
    def cancel(self):
        pass

    def show_status(self):  # متد معمولی
        status = "Booked" if self.is_booked else "Available"
        return f"{self.customer_name}: {status}"

class HotelReservation(Reservation):
    def __init__(self, customer_name, room_number):
        super().__init__(customer_name)
        self.room_number = room_number

    def book(self):
        self.is_booked = True
        return f"Room {self.room_number} booked for {self.customer_name}"

    def cancel(self):
        self.is_booked = False
        return f"Room {self.room_number} cancelled"

class FlightReservation(Reservation):
    def __init__(self, customer_name, flight_number):
        super().__init__(customer_name)
        self.flight_number = flight_number

    def book(self):
        self.is_booked = True
        return f"Flight {self.flight_number} booked for {self.customer_name}"

    def cancel(self):
        self.is_booked = False
        return f"Flight {self.flight_number} cancelled"

hotel = HotelReservation("Amir", 40)
flight = FlightReservation("Sara", "IR123")
print(hotel.book())    # Room 40 booked for Amir
print(flight.book())   # Flight IR123 booked for Sara
print(hotel.show_status()) # Amir: Booked
```

# Composition


ترکیب یک اصل بنیادی در برنامه‌نویسی شی‌گرا است که در آن اشیاء پیچیده از اشیاء ساده‌تر ساخته می‌شوند. در این رویکرد، کلاس‌ها به جای ارث‌بری از یکدیگر، از اجزای مستقل تشکیل می‌شوند که هر کدام رفتار خاص خود را دارند. ترکیب به عنوان "رابطه هست-یک" در مقابل وراثت که "رابطه است-یک" است، شناخته می‌شود. این اصل باعث افزایش انعطاف‌پذیری، قابلیت استفاده مجدد و کاهش وابستگی در طراحی نرم‌افزار می‌شود.

Composition in object-oriented programming is a design principle where a complex object is constructed from simpler, independent objects. It represents a "has-a" relationship, where one class contains instances of other classes as its components, rather than inheriting from them.

```python
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        self.engine = Engine()  # ترکیب‌سازی: ماشین یک موتور دارد

    def start(self):
        return self.engine.start()  # استفاده از موتور

my_car = Car()
print(my_car.start())
```

```python
class Pen:
    def write(self):
        return "Writing with pen"

class Student:
    def __init__(self, name):
        self.name = name
        self.pen = Pen()  # ترکیب‌سازی: دانشجو یک خودکار دارد

    def study(self):
        return f"{self.name} is {self.pen.write()}"


amir = Student("Amir")
print(amir.study())  # Amir is Writing with pen
```

```python
class Book:
    def __init__(self, title):
        self.title = title

    def read(self):
        return f"Reading {self.title}"

class Library:
    def __init__(self):
        self.books = []  # ترکیب‌سازی: کتابخانه چندین کتاب دارد

    def add_book(self, book):
        self.books.append(book)
        return f"Added {book.title}"

    def show_books(self):
        return [book.title for book in self.books]


library = Library()
book1 = Book("Python 101")
book2 = Book("Data Science")
library.add_book(book1)  # Added Python 101
library.add_book(book2)  # Added Data Science
print(library.show_books())  # ['Python 101', 'Data Science']
```

```python
class CPU:
    def process(self):
        return "Processing data"

class RAM:
    def load(self):
        return "Loading data"

class Storage:
    def save(self):
        return "Saving data"

class Computer:
    def __init__(self):
        self.cpu = CPU()      # ترکیب‌سازی
        self.ram = RAM()      # ترکیب‌سازی
        self.storage = Storage()  # ترکیب‌سازی

    def run(self):
        return f"{self.ram.load()} -> {self.cpu.process()} -> {self.storage.save()}"


pc = Computer()
print(pc.run())  # Loading data -> Processing data -> Saving data
```

```python
class Chef:
    def __init__(self, name):
        self.name = name

    def cook(self, dish):
        return f"Chef {self.name} is cooking {dish}"

class Waiter:
    def __init__(self, name):
        self.name = name

    def serve(self, dish):
        return f"Waiter {self.name} is serving {dish}"

class Restaurant:
    def __init__(self, chef_name, waiter_names):
        self.chef = Chef(chef_name)  # ترکیب‌سازی
        self.waiters = [Waiter(name) for name in waiter_names]  # ترکیب‌سازی

    def order(self, dish):
        result = self.chef.cook(dish)
        for waiter in self.waiters:
            result += f"\n{waiter.serve(dish)}"
        return result


restaurant = Restaurant("Ali", ["Mohana", "Mahbod"])
print(restaurant.order("Pizza"))
```

```python
class Course:
    def __init__(self, name, teacher):
        self.name = name
        self.teacher = teacher

    def show_info(self):
        return f"Course: {self.name} (Teacher: {self.teacher})"

class Student:
    def __init__(self, name):
        self.name = name
        self.courses = []  # ترکیب‌سازی

    def add_course(self, course):
        self.courses.append(course)
        return f"{self.name} added {course.name}"

    def show_courses(self):
        return [course.show_info() for course in self.courses]

# تست
math = Course("Math 101", "Dr. Smith")
physics = Course("Physics 202", "Dr. Johnson")
amir = Student("Amir")
amir.add_course(math)   # Amir added Math 101
amir.add_course(physics) # Amir added Physics 202
print(amir.show_courses())
```

ترکیب سازی با کپسوله سازی

```python
class Date:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    def show(self):
        return f"{self.year}-{self.month:02d}-{self.day:02d}"

class Person:
    def __init__(self, name, year, month, day):
        self.name = name
        self.__birthday = Date(year, month, day)  # ترکیب‌سازی + کپسوله‌سازی

    def get_age(self):
        current_year = 2024
        return current_year - self.__birthday.year

    def show_birthday(self):
        return self.__birthday.show()

amir = Person("Amir", 2000, 5, 15)
print(amir.show_birthday())  # 2000-05-15
print(amir.get_age())        # 24
```

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def work(self):
        return f"{self.name} is working as {self.role}"

class Department:
    def __init__(self, name):
        self.name = name
        self.employees = []  # ترکیب‌سازی

    def add_employee(self, employee):
        self.employees.append(employee)
        return f"Added {employee.name} to {self.name}"

    def show_employees(self):
        return [emp.work() for emp in self.employees]

class Company:
    def __init__(self, name):
        self.name = name
        self.departments = []  # ترکیب‌سازی

    def add_department(self, department):
        self.departments.append(department)
        return f"Added {department.name} to {self.name}"

    def show_all(self):
        result = f"Company: {self.name}\n"
        for dept in self.departments:
            result += f"\nDepartment: {dept.name}\n"
            for emp in dept.show_employees():
                result += f"  - {emp}\n"
        return result


emp1 = Employee("Amir", "Developer")
emp2 = Employee("Hasti", "Manager")
emp3 = Employee("Parsa", "Designer")

it_dept = Department("IT")
hr_dept = Department("HR")

it_dept.add_employee(emp1)
it_dept.add_employee(emp2)
hr_dept.add_employee(emp3)

company = Company("TechCorp")
company.add_department(it_dept)
company.add_department(hr_dept)

print(company.show_all())
```
